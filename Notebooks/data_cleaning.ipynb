```python
#importing both the dataset
dataset_1 = r"C:\Users\Admin\Desktop\CP\Data\raw\Dataset_1.xlsx"
dataset_2 = r"C:\Users\Admin\Desktop\CP\Data\raw\Dataset_2.xlsx"
df_1_hip = pd.read_excel(dataset_1,sheet_name="Hip")
df_1_knee = pd.read_excel(dataset_1,sheet_name="Knee")
df_2 = pd.read_excel(dataset_2,sheet_name='ALL')
```

```python
hip_angle_column = []
knee_angle_column = []
for i in range(0,1001,20):
    hip_angle_column.append(f"aSagH_pct_GC_{i}")
    knee_angle_column.append(f"aSagK_pct_GC_{i}")
final = hip_angle_column+knee_angle_column
final += ["Patient ID","side",]
df_2 = df_2[final]
```

```python
df_2
```

```python
# Average per patient per side
df_side_avg = df_2.groupby(['Patient ID', 'side']).mean().reset_index()
import re

def rename_columns(col):
    match = re.match(r'(aSag\w+)_pct_GC_(\d+)', col)
    if match:
        joint, pct = match.groups()
        return f'{joint}_{int(pct) // 10}'
    else:
        return col  # return unchanged if it doesn't match pattern

# Apply to all columns
df_side_avg.rename(columns=rename_columns, inplace=True)
```
